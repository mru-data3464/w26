```python
# a sequence of bytes
with open("bytes.txt", "wb") as f:
    for b in range(256):
        f.write(b.to_bytes())
```

```python
# read it back
with open("bytes.txt", "rb") as f:
    all_bytes = f.read()

all_bytes
```

```python
# just the printable character range
printable = range(32, 127)
with open("printable.txt", "wb") as f:
    for b in printable:
        f.write(b.to_bytes())
```

```python
# creating bytes
print((0).to_bytes())
print(b"HI")
for b in b"HI":
    print(bin(b), end=" ")
```

```python
# Files in different encodings
with open("utf8.txt", "w", encoding="utf-8", ) as f:
    f.write("ABC 🔥")

with open("utf16.txt", "w", encoding="utf-16") as f:
    f.write("ABC 🔥")
```

```python
# read back the bytes
print("Utf-8:")
with open("utf8.txt", "rb") as f:
    for b in f.read():
        print(b.to_bytes())

print("\n\nUtf-16:")
with open("utf16.txt", "rb") as f:
    for b in f.read():
        print(b.to_bytes())
```

```python
# What happens if we use the wrong encoding?
with open("utf16.txt", "r") as f:
    print(f.read())
```
