```python
# a sequence of bytes
with open("bytes.txt", "wb") as f:
    for b in range(256):
        f.write(b.to_bytes())
```

```python
# read it back
with open("bytes.txt", "rb") as f:
    all_bytes = f.read()

all_bytes

for b in all_bytes:
    print(bin(b))
```

```python
# just the printable character range
printable = range(32, 127)
with open("printable.txt", "wb") as f:
    for b in printable:
        f.write(b.to_bytes())
```

```python
# creating bytes
print((0).to_bytes())
print(b"HI")
for b in b"HI":
    print(bin(b), end=" ")
```

```python
# Files in different encodings
with open("utf8.txt", "w", encoding="utf-8", ) as f:
    f.write("ABC 🔥")

with open("utf16.txt", "w", encoding="utf-16") as f:
    f.write("ABC 🔥")
```

```python
# read back the bytes
print("Utf-8:")
with open("utf8.txt", "rb") as f:
    for b in f.read():
        print(b.to_bytes())

print("\n\nUtf-16:")
with open("utf16.txt", "rb") as f:
    for b in f.read():
        print(b.to_bytes())
```

# More regex adventures

```python
# raw string demo
# print("newline\nnext line")
# print(r"newline\nnext line")

import re
test_string = """
$123
$10,000
$1,000,000.00
$36.50
1234
1,000,000
$no numbers at all
"""

# python's "raw" string means we don't have to double escape things
matches = re.finditer(r"\$(\d+[,\d{3}]*(?:\.\d\d)?)", test_string)
for m in matches:
    print(m)
    print(m.group(1))
```

```python
phone_numbers = """
a bunch of things
some of them are phone numbers: 555-555-1234
Sometimes they don't have the hyphen: 123 345 1234
I don't want to match any 3 digits like 404 or 555
"""
matches = re.finditer(r"(\d{3})[-\s]\d{3}[-\s]\d{4}", phone_numbers)
area_codes = []
for m in matches:
    print(m, m.group(1))
    area_codes.append(int(m.group(1)))

print(area_codes)
```

```python
emails = """
someone@mtroyal.ca
else@mru.ca
maybe@gmail.com
some_people_still_use@hotmail.com
assume it's not just emails
maybe there's a stray @symbol
something@mtroyal
"""
# goal: retrieve usernames and domain from emails
# might be simplifying email rules a bit
matches = re.finditer(r"([a-zA-Z]\w*)\@(\w+\.[a-z]+)", emails)
usernames = []
domains = []
for m in matches:
    usernames.append(m.group(1))
    domains.append(m.group(2))

print(usernames)
print(domains)
```
