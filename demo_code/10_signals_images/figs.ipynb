```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generate a sample "continuous" signal
# Two constant frequencies
f1 = 15 # Hz
f2 = 60 # Hz

# and phase shifts
p1 = 0
p2 = np.pi / 4

# and amplitudes
a1 = 1
a2 = 3

# Time interval
T = 1 / (f2 * 20)

# period is the greatest common denominator of the two frequencies
P = 1 / np.gcd(int(f1), int(f2))

# Number of samples
N = int(P / T)

t = np.arange(N) * T
x = a1 * np.cos(f1 * 2*np.pi*t + p1) + a2 *np.cos(f2 * 2*np.pi*t + p2)

plt.plot(t, x, "k--", label="Continuous signal")
plt.stem(t, x, label="Discrete signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.legend()
```

```python
X = 2 * np.fft.fft(x) / N
k = np.fft.fftfreq(N, d=T)

plt.stem(k, abs(X))
plt.text(f1, a1 + 0.1, "$f_1$")
plt.text(f2, a2 + 0.1, "$f_2$")
plt.xlim([0,200])
```

```python
# Phase
mask = abs(X) > 0.1
plt.stem(k[mask], np.angle(X[mask]))
plt.xlim([0, f2 + f1])
plt.xlabel("Frequency")
plt.ylabel("Phase")
```

## Audio Data
Samples downloaded from [AudioMNIST](https://github.com/soerenab/AudioMNIST/tree/master).

```python
from scipy.io import wavfile
import IPython.display as ipd
```

```python
# Load the two audio files
br_00, zero_0 = wavfile.read("0_01_0.wav")
br_1, zero_1 = wavfile.read("0_01_1.wav")

# what did that return?
print(br_00, zero_0.shape)
print(br_1, zero_1.shape)
```

```python
t = np.arange(len(zero_0)) / br_00
plt.plot(t, zero_0)
```

```python
# listen to it
ipd.Audio(zero_0, rate=br_00)
```

```python
# How about a single tone?
tone = np.cos(440 * 2*np.pi*t)
ipd.Audio(tone, rate=br_00)
```

```python
# what does the frequency version look like?
def plot_freq(audio, br, max_f):
    N = len(audio)
    F = np.fft.fft(audio) * 2 / N
    k = np.fft.fftfreq(N, d=1/br)
    plt.plot(k[:N//2], abs(F[:N//2]))
    plt.xlabel("Frequency (Hz)")
    plt.ylabel("Amplitude")
    plt.xlim([0, max_f])

plot_freq(tone, br_00, 1000)
```

```python
# What about the zero?
plot_freq(zero_0, br_00, 2000)
```

```python
# What if we truncate and bring back to time?
zero_F = np.fft.fft(zero_0) * 2 / N
k = np.fft.fftfreq(len(zero_0), d=1/br_00)
new_br = 4000
lp_zero = np.real(np.fft.ifft(zero_F[abs(k) < new_br / 2]))
plt.plot(lp_zero)

# What does it sound like?
ipd.Audio(lp_zero, rate=new_br)
```

```python
# just for fun, mash the two channels together
N = min(len(zero_0), len(zero_1))
two_chan = np.vstack([zero_0[:N], zero_1[:N]])
two_chan.shape

plt.plot(t[:N], two_chan.T)

ipd.Audio(two_chan, rate=br_00)
```
