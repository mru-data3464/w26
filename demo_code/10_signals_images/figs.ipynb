```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generate a sample "continous" signal
# Two fundamental frequencies
f1 = 15 # Hz
f2 = 60 # Hz

# and phase shifts
p1 = 0
p2 = np.pi / 4

# and amplitudes
a1 = 1
a2 = 3

# Number of samples
N = 100

# sampling rate
dt = 1 / (f2 * 20)

t = np.linspace(0, N*dt, N)
y = a1 * np.sin(f1 * 2*np.pi*t + p1) + a2 *np.sin(f2 * 2*np.pi*t + p2)

plt.plot(t, y, "k--", label="Continuous signal")
plt.stem(t, y, label="Discrete signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.legend()
```

```python
Y = 2 * np.fft.fft(y) / N
maxfreq = 1/dt / 2
k = np.concat([np.linspace(0, maxfreq, N//2), np.linspace(-maxfreq, 0, N // 2)])

plt.stem(k, abs(Y))
plt.text(f1, a1 + 0.1, "$f_1$")
plt.text(f2, a2 + 0.1, "$f_2$")
plt.xlim([0,max(k)])
```

```python
# Phase
plt.stem(k, np.angle(Y))
plt.xlim([0,max(k)])
plt.xlabel("Frequency")
plt.ylabel("Phase")
```
