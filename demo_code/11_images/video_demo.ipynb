```python
import numpy as np
import matplotlib.pyplot as plt
import cv2 as cv
```

```python
cap = cv.VideoCapture("vtest.avi")
```

```python
# Video properties listed at https://docs.opencv.org/4.13.0/d4/d15/group__videoio__flags__base.html#gaeb8dd9c89c10a5c63c139bf7c4f5704d

props = {
    "width": cv.CAP_PROP_FRAME_WIDTH,
    "height": cv.CAP_PROP_FRAME_HEIGHT,
    "frame rate": cv.CAP_PROP_FPS,
    "frames": cv.CAP_PROP_FRAME_COUNT,
}

for k, v in props.items():
    print(f"{k}: {cap.get(props[k])}")
```

```python
# read and display frames
_, f1 = cap.read()
f1
```

```python
plt.imshow(f1)
```

```python
# next
_, f2 = cap.read()
plt.imshow(f2)
```

```python
# difference? Not so simple...
plt.imshow(abs(f2 - f1))
```

```python
# some funky integer behaviour going on
print(f2[-1,-1,:])
print(f1[-1,-1,:])
print(abs(f2 - f1)[-1,-1,:])
```

```python
# convert to floats and rescale
f1_f = f1 / 255
f2_f = f2 / 255
plt.imshow(abs(f2_f - f1_f))
```
